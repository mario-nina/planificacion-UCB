# A. IK y Pieper — Desacoplamiento

### Consigna

Dada una pose deseada:

$$
{}^0T_{6,d}=
\begin{bmatrix}
R_d & p_d\\
0 & 1
\end{bmatrix}
$$

con:

$$
p_d=
\begin{bmatrix}
0.8\\
0\\
0.4
\end{bmatrix}
$$

y una orientación tal que:

$$
{}^0z_6^d=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
$$

Para un offset final de:

$$
d_6=0.1
$$

resuelva:

1. Calcule el centro de muñeca $p_W$.
2. Si una resolución parcial del brazo nos lleva a:

$$
c_2=0.5
\qquad\Rightarrow\qquad
s_2=\pm\sqrt{0.75}
$$

construya la expresión rigurosa de recuperación del ángulo utilizando $\operatorname{atan2}$.

## Resolución

### 1. Cálculo del centro de muñeca

Para el desacoplamiento cinemático de un manipulador con muñeca esférica, el centro de muñeca se obtiene restando de la posición deseada el desplazamiento final $d_6$ en la dirección del eje $z_6$:

$$
p_W=p_d-d_6\,{}^0z_6^d
$$

Los datos proporcionados son:

$$
p_d=
\begin{bmatrix}
0.8\\
0\\
0.4
\end{bmatrix}
$$

$$
{}^0z_6^d=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
$$

y:

$$
d_6=0.1
$$

Primero calculamos el desplazamiento correspondiente al offset:

$$
d_6\,{}^0z_6^d
=
0.1
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
$$

Por lo tanto:

$$
d_6\,{}^0z_6^d
=
\begin{bmatrix}
0.1\\
0\\
0
\end{bmatrix}
$$

Ahora calculamos el centro de muñeca:

$$
p_W
=
\begin{bmatrix}
0.8\\
0\\
0.4
\end{bmatrix}
-
\begin{bmatrix}
0.1\\
0\\
0
\end{bmatrix}
$$

Entonces:

$$
\boxed{
p_W=
\begin{bmatrix}
0.7\\
0\\
0.4
\end{bmatrix}
}
$$

El centro de muñeca se encuentra, por tanto, a $0.7$ unidades en $x$ y $0.4$ unidades en $z$.

### 2. Recuperación del ángulo mediante $\operatorname{atan2}$

La resolución parcial proporciona:

$$
c_2=\cos(q_2)=0.5
$$

y:

$$
s_2=\sin(q_2)=\pm\sqrt{0.75}
$$

Simplificamos:

$$
\sqrt{0.75}
=
\sqrt{\frac{3}{4}}
=
\frac{\sqrt{3}}{2}
$$

Por lo tanto:

$$
s_2=\pm\frac{\sqrt{3}}{2}
$$

Para recuperar el ángulo $q_2$ utilizamos la función de dos argumentos:

$$
\boxed{
q_2=\operatorname{atan2}(s_2,c_2)
}
$$

Sustituyendo los valores:

$$
q_2=
\operatorname{atan2}
\left(
\pm\frac{\sqrt{3}}{2},
0.5
\right)
$$

Esto genera dos posibles ramas.

### Rama positiva

Para:

$$
s_2=+\frac{\sqrt{3}}{2}
$$

tenemos:

$$
q_2=
\operatorname{atan2}
\left(
\frac{\sqrt{3}}{2},
\frac{1}{2}
\right)
$$

por lo que:

$$
\boxed{
q_2=\frac{\pi}{3}
}
$$

### Rama negativa

Para:

$$
s_2=-\frac{\sqrt{3}}{2}
$$

tenemos:

$$
q_2=
\operatorname{atan2}
\left(
-\frac{\sqrt{3}}{2},
\frac{1}{2}
\right)
$$

por lo que:

$$
\boxed{
q_2=-\frac{\pi}{3}
}
$$

Así, las dos soluciones posibles son:

$$
\boxed{
q_2=\pm\frac{\pi}{3}
}
$$

## Resultado

El centro de muñeca obtenido es:

$$
\boxed{
p_W=
\begin{bmatrix}
0.7\\
0\\
0.4
\end{bmatrix}
}
$$

La recuperación del ángulo debe realizarse mediante:

$$
\boxed{
q_2=\operatorname{atan2}(s_2,c_2)
}
$$

con:

$$
c_2=0.5,
\qquad
s_2=\pm\frac{\sqrt{3}}{2}
$$

Por lo tanto, las dos ramas posibles son:

$$
\boxed{
q_2=+\frac{\pi}{3}
}
$$

y

$$
\boxed{
q_2=-\frac{\pi}{3}
}
$$

El uso de $\operatorname{atan2}$ permite conservar correctamente la información de los signos de seno y coseno y, por tanto, distinguir las diferentes ramas de la solución.

# B. Jacobiano

### Consigna

Evalúe la siguiente construcción hipotética en:

$$
q=
\begin{bmatrix}
0\\
\frac{\pi}{2}
\end{bmatrix}
$$

Los ejes $Z$ coinciden con el vector global $Z_0$:

$$
J=
\begin{bmatrix}
\cdots & -0.5\\
\cdots & 0\\
0 & 0\\
0 & 0\\
0 & 0\\
1 & 1
\end{bmatrix}
$$

Si se requiere:

$$
\dot q=
\begin{bmatrix}
0\\
2
\end{bmatrix}
$$

calcule la velocidad espacial del efector final:

$$
v_e=J\dot q
$$

## Resolución

La velocidad espacial del efector final se obtiene mediante:

$$
v_e=J\dot q
$$

Tenemos:

$$
\dot q=
\begin{bmatrix}
0\\
2
\end{bmatrix}
$$

y:

$$
J=
\begin{bmatrix}
\cdots & -0.5\\
\cdots & 0\\
0 & 0\\
0 & 0\\
0 & 0\\
1 & 1
\end{bmatrix}
$$

Al realizar la multiplicación:

$$
v_e=
\begin{bmatrix}
\cdots & -0.5\\
\cdots & 0\\
0 & 0\\
0 & 0\\
0 & 0\\
1 & 1
\end{bmatrix}
\begin{bmatrix}
0\\
2
\end{bmatrix}
$$

La multiplicación de una matriz por un vector puede interpretarse como una combinación lineal de las columnas del Jacobiano:

$$
v_e
=
0J_1+2J_2
$$

Como:

$$
0J_1=0
$$

la primera columna no aporta ninguna contribución a la velocidad.

Por lo tanto, únicamente necesitamos la segunda columna:

$$
J_2=
\begin{bmatrix}
-0.5\\
0\\
0\\
0\\
0\\
1
\end{bmatrix}
$$

Multiplicando por la velocidad de la segunda articulación:

$$
2J_2
=
2
\begin{bmatrix}
-0.5\\
0\\
0\\
0\\
0\\
1
\end{bmatrix}
$$

obtenemos:

$$
v_e=
\begin{bmatrix}
-1\\
0\\
0\\
0\\
0\\
2
\end{bmatrix}
$$

La velocidad espacial está formada por una parte lineal y una parte angular:

$$
v_e=
\begin{bmatrix}
v_L\\
\omega
\end{bmatrix}
$$

Por lo tanto:

$$
v_L=
\begin{bmatrix}
-1\\
0\\
0
\end{bmatrix}
$$

y:

$$
\omega=
\begin{bmatrix}
0\\
0\\
2
\end{bmatrix}
$$

## Resultado

La velocidad espacial del efector final es:

$$
\boxed{
v_e=
\begin{bmatrix}
-1\\
0\\
0\\
0\\
0\\
2
\end{bmatrix}
}
$$

Separando sus componentes:

### Velocidad lineal

\boxed{

$$

Esto significa que, para la velocidad articular indicada, el efector final presenta una velocidad lineal de $-1$ en la dirección $X$ y una velocidad angular de $2$ alrededor del eje $Z$.

La primera articulación no contribuye en este instante porque:

$$
\dot q_1=0

$$
\dot q_2=2

# C. Singularidades

### Consigna

Para un sub-Jacobiano:

$$
J_p=
\begin{bmatrix}
-2s_{12}-s_1 & -2s_{12}\\
2c_{12}+c_1 & 2c_{12}
\end{bmatrix}
$$

donde:

$$
s_{12}=\sin(q_1+q_2)
$$

y

$$
c_{12}=\cos(q_1+q_2)
$$

demuestre analíticamente si la configuración:

$$
q_2=\pi
$$

produce una pérdida de rango.

Identifique la dirección de movimiento cartesiano que se pierde.

## Resolución

Para determinar si existe una singularidad, debemos analizar el rango del sub-Jacobiano.

Como $J_p$ es una matriz $2\times2$, existe pérdida de rango cuando:

$$
\det(J_p)=0
$$

Partimos de:

$$
J_p=
\begin{bmatrix}
-2s_{12}-s_1 & -2s_{12}\\
2c_{12}+c_1 & 2c_{12}
\end{bmatrix}
$$

### 1. Evaluamos en $q_2=\pi$

Cuando:

$$
q_2=\pi
$$

tenemos:

$$
s_{12}=\sin(q_1+\pi)
$$

Como:

$$
\sin(q_1+\pi)=-\sin q_1
$$

entonces:

$$
s_{12}=-s_1
$$

También:

$$
c_{12}=\cos(q_1+\pi)
$$

y:

$$
\cos(q_1+\pi)=-\cos q_1
$$

por lo que:

$$
c_{12}=-c_1
$$

Sustituyendo estas relaciones en el Jacobiano:

$$
J_p(q_2=\pi)=
\begin{bmatrix}
-2(-s_1)-s_1 & -2(-s_1)\\
2(-c_1)+c_1 & 2(-c_1)
\end{bmatrix}
$$

Simplificando:

$$
J_p(q_2=\pi)=
\begin{bmatrix}
s_1 & 2s_1\\
-c_1 & -2c_1
\end{bmatrix}
$$

---

### 2. Calculamos el determinante

El determinante de una matriz $2\times2$:

$$
\begin{bmatrix}
a&b\\
c&d
\end{bmatrix}
$$

es:

$$
\det(J_p)=ad-bc
$$

Por tanto:

$$
\det(J_p)
=
s_1(-2c_1)
-
(2s_1)(-c_1)
$$

$$
\det(J_p)
=
-2s_1c_1+2s_1c_1
$$

Entonces:

$$
\boxed{
\det(J_p)=0
}
$$

Por lo tanto, en $q_2=\pi$ las columnas del Jacobiano son linealmente dependientes y el rango disminuye.

$$
\boxed{
\operatorname{rank}(J_p)<2
}
$$

### 3. Identificación de la dirección perdida

Las dos columnas del Jacobiano son:

$$
J_{p,1}=
\begin{bmatrix}
s_1\\
-c_1
\end{bmatrix}
$$

y

$$
J_{p,2}=
\begin{bmatrix}
2s_1\\
-2c_1
\end{bmatrix}
$$

Observamos que:

$$
J_{p,2}=2J_{p,1}
$$

Por lo tanto, ambas columnas son linealmente dependientes y el Jacobiano solamente puede generar movimiento instantáneo en una dirección del plano.

La dirección que sí puede generar movimiento es:

$$
\begin{bmatrix}
s_1\\
-c_1
\end{bmatrix}
$$

Una dirección perpendicular a esta es:

$$
\begin{bmatrix}
c_1\\
s_1
\end{bmatrix}
$$

Esta es precisamente la dirección cartesiana que no puede generarse instantáneamente en la configuración singular.

Por lo tanto, la dirección perdida es:

$$
\boxed{
\begin{bmatrix}
c_1\\
s_1
\end{bmatrix}
}
$$

Esta dirección corresponde a la dirección radial asociada con la configuración del brazo.

## Resultado

Para:

$$
q_2=\pi
$$

se obtiene:

$$
\boxed{
\det(J_p)=0
}
$$

y, por tanto:

$$
\boxed{
\operatorname{rank}(J_p)=1<2
}
$$

Por lo tanto, **sí existe una pérdida de rango y la configuración es singular**.

Las columnas del Jacobiano son dependientes:

$$
J_{p,2}=2J_{p,1}
$$

La dirección cartesiana que se pierde es:

$$
\boxed{
\begin{bmatrix}
c_1\\
s_1
\end{bmatrix}
}
$$

Es decir, se pierde la capacidad de producir movimiento instantáneo en la dirección radial del brazo.

# D. Trayectoria LSPB

### Consigna

Debe moverse de:

$$
q_i=0
$$

a:

$$
q_f=4
$$

en:

$$
t_f=4\ \text{s}
$$

¿Es factible configurar un perfil trapezoidal con:

$$
\ddot{q}_c=1\ \text{rad/s}^2
$$

?

De ser así, determine el tiempo de mezcla de aceleración $t_c$ e interprete las fases del movimiento.

## Resolución

Para un perfil LSPB simétrico, el movimiento está compuesto por tres fases:

1. **Aceleración**
2. **Velocidad constante**
3. **Desaceleración**

La aceleración tiene magnitud:

$$
\ddot{q}_c=1\ \text{rad/s}^2
$$

y el tiempo total disponible es:

$$
t_f=4\ \text{s}
$$

El desplazamiento total es:

$$
\Delta q=q_f-q_i
$$

por lo que:

$$
\Delta q=4-0=4\ \text{rad}
$$

---

### 1. Relación del desplazamiento con $t_c$

Durante la primera fase, la velocidad aumenta desde cero hasta:

$$
\dot q_c=\ddot q_c t_c
$$

Durante la fase de crucero se mantiene esa velocidad.

Por simetría, la desaceleración también dura $t_c$.

Por tanto, la duración de la fase de velocidad constante es:

$$
t_f-2t_c
$$

El desplazamiento total de un perfil trapezoidal simétrico es:

$$
\Delta q
=
\ddot q_c t_c(t_f-t_c)
$$

Sustituyendo los valores:

$$
4
=
(1)t_c(4-t_c)
$$

Por lo tanto:

$$
4=4t_c-t_c^2
$$

Reordenando:

$$
t_c^2-4t_c+4=0
$$

Factorizamos:

$$
(t_c-2)^2=0
$$

Entonces:

$$
\boxed{
t_c=2\ \text{s}
}
$$

---

### 2. Verificación de factibilidad

La fase de velocidad constante tendría una duración:

$$
t_f-2t_c
=
4-2(2)
$$

por lo que:

$$
t_f-2t_c=0
$$

Esto significa que **no existe un tramo de velocidad constante de duración positiva**.

El perfil llega exactamente al límite entre un perfil trapezoidal y un perfil triangular.

La velocidad máxima alcanzada sería:

$$
\dot q_c=\ddot q_c t_c
$$

$$
\dot q_c=(1)(2)
$$

por lo tanto:

$$
\boxed{
\dot q_c=2\ \text{rad/s}
}
$$


## Resultado

El tiempo de aceleración obtenido es:

$$
\boxed{
t_c=2\ \text{s}
}
$$

Sin embargo:

$$
t_f-2t_c=0
$$

por lo que **no existe una fase de velocidad de crucero**.

El movimiento queda compuesto por:

- **Aceleración:** $0\leq t\leq2\ \text{s}$
- **Desaceleración:** $2\leq t\leq4\ \text{s}$
- **Velocidad constante:** duración $0\ \text{s}$

La velocidad máxima es:

$$
\boxed{
\dot q_{\max}=2\ \text{rad/s}
}
$$

Por lo tanto, con los valores dados, el movimiento es factible **como caso límite de un perfil LSPB**, pero no como un perfil trapezoidal con una fase de crucero de duración positiva. En la práctica, corresponde a un perfil triangular.

# E. Dinámica de 1R

### Consigna

Evaluando un eslabón 1R en el momento pico de la fase de aceleración calculada en LSPB, considerando:

$$
J_{eq}=0.5\ \text{kg·m}^2
$$

$$
\ell_c=0.5\ \text{m}
$$

$$
m=2\ \text{kg}
$$

calcule el torque requerido en la configuración instantánea:

$$
q=0.5\ \text{rad}
$$

Identifique la contribución inercial y la gravitacional.

## Resolución

Para el modelo dinámico 1R utilizamos:

$$
\tau =
J_{eq}\ddot q
+
mg\ell_c\sin q
$$

La ecuación está formada por dos contribuciones:

$$
\tau=
\underbrace{J_{eq}\ddot q}_{\text{inercial}}
+
\underbrace{mg\ell_c\sin q}_{\text{gravitacional}}
$$

### 1. Aceleración en el instante considerado

En la fase de aceleración del perfil LSPB se tiene:

$$
\ddot q_c=1\ \text{rad/s}^2
$$

Por lo tanto, en el momento pico de aceleración:

$$
\ddot q=1\ \text{rad/s}^2
$$

---

### 2. Contribución inercial

La contribución inercial es:

$$
\tau_I=J_{eq}\ddot q
$$

Sustituyendo:

$$
\tau_I=(0.5)(1)
$$

Por lo tanto:

$$
\boxed{
\tau_I=0.5\ \text{N·m}
}
$$

---

### 3. Contribución gravitacional

La contribución gravitacional es:

$$
\tau_g=mg\ell_c\sin q
$$

Sustituimos:

$$
m=2\ \text{kg}
$$

$$
g=9.81\ \text{m/s}^2
$$

$$
\ell_c=0.5\ \text{m}
$$

y:

$$
q=0.5\ \text{rad}
$$

Entonces:

$$
\tau_g
=
(2)(9.81)(0.5)\sin(0.5)
$$

Primero:

$$
(2)(9.81)(0.5)=9.81
$$

y:

$$
\sin(0.5)\approx0.4794
$$

Por tanto:

$$
\tau_g
=
9.81(0.4794)
$$

$$
\boxed{
\tau_g\approx4.703\ \text{N·m}
}
$$

---

### 4. Torque total

El torque total es la suma de ambas contribuciones:

$$
\tau=\tau_I+\tau_g
$$

Sustituyendo:

$$
\tau
=
0.5+4.703
$$

Por lo tanto:

$$
\boxed{
\tau\approx5.203\ \text{N·m}
}
$$

## Resultado

Las contribuciones del torque son:

### Contribución inercial

$$
\boxed{
\tau_I=0.5\ \text{N·m}
}
$$

### Contribución gravitacional

$$
\boxed{
\tau_g\approx4.703\ \text{N·m}
}
$$

### Torque total requerido

$$
\boxed{
\tau\approx5.203\ \text{N·m}
}
$$

Por lo tanto, aunque el momento considerado corresponde al pico de aceleración, la contribución gravitacional es mucho mayor que la contribución inercial en esta configuración.

La relación entre ambas contribuciones es:

$$
\tau_g>\tau_I
$$

# F. Razonamiento Físico Integrado

### Consigna

Dado el estado instantáneo calculado en la sección E, justifique simultáneamente:

**¿Por qué hay alto riesgo dinámico pero bajo riesgo de singularidad en ese punto de la trayectoria?**

## Resolución

Para analizar esta situación debemos separar dos conceptos diferentes:

1. El **riesgo dinámico**, relacionado con el esfuerzo que deben realizar los actuadores.
2. El **riesgo de singularidad**, relacionado con la movilidad instantánea del robot y el rango de su Jacobiano.

---

### 1. ¿Por qué existe un alto riesgo dinámico?

En la sección E obtuvimos un torque total aproximado de:

$$
\tau\approx5.203\ \text{N·m}
$$

Este torque está compuesto por:

$$
\tau=\tau_I+\tau_g
$$

donde:

$$
\tau_I=0.5\ \text{N·m}
$$

es la contribución inercial y:

$$
\tau_g\approx4.703\ \text{N·m}
$$

es la contribución gravitacional.

La mayor contribución corresponde a la gravedad:

$$
\tau_g>\tau_I
$$

Por lo tanto, el actuador debe realizar un esfuerzo considerable para sostener y acelerar el eslabón en esa configuración.

Además, el estado corresponde al momento de máxima aceleración de la fase considerada:

$$
\ddot q=1\ \text{rad/s}^2
$$

por lo que existe simultáneamente una contribución inercial.

En consecuencia, existe un **riesgo dinámico elevado**, debido principalmente a la demanda de torque gravitacional y, adicionalmente, a la aceleración.

---

### 2. ¿Por qué el riesgo de singularidad es bajo?

Una singularidad cinemática no se determina observando el torque.

La condición de singularidad se establece mediante el Jacobiano:

$$
\operatorname{rank}(J(q))
<
\operatorname{rank}_{\max}
$$

Es decir, existe una singularidad cuando se produce una pérdida de rango y, por tanto, una pérdida de movilidad cartesiana instantánea.

En cambio, un torque elevado puede aparecer simplemente debido a:

$$
\tau=
M(q)\ddot q
+
C(q,\dot q)\dot q
+
g(q)
$$

Por eso:

$$
\boxed{
\text{Torque alto}
\not\Rightarrow
\text{Singularidad}
}
$$

En el estado analizado en la sección E no se ha demostrado una pérdida de rango del Jacobiano. Por lo tanto, con la información proporcionada por la práctica, no podemos afirmar que exista una singularidad en ese punto.

---

### 3. Interpretación conjunta

El estado estudiado presenta dos situaciones diferentes:

**Dinámica:**

$$
\tau\approx5.203\ \text{N·m}
$$

por lo que existe una demanda importante de esfuerzo del actuador.

**Cinemática:**

La existencia de una singularidad tendría que comprobarse mediante:

$$
\operatorname{rank}(J(q))
$$

y no mediante el valor del torque.

Por lo tanto, un punto de la trayectoria puede presentar una demanda dinámica elevada y, al mismo tiempo, permanecer alejado de una singularidad cinemática.

## Resultado

Existe un **alto riesgo dinámico** porque el actuador debe proporcionar un torque aproximado de:

$$
\boxed{
\tau\approx5.203\ \text{N·m}
}
$$

con contribuciones:

$$
\boxed{
\tau_I=0.5\ \text{N·m}
}
$$

y:

$$
\boxed{
\tau_g\approx4.703\ \text{N·m}
}
$$

La principal demanda proviene de la gravedad.

Por otro lado, existe un **bajo riesgo de singularidad** porque un torque elevado no implica una singularidad. La singularidad debe determinarse mediante la pérdida de rango del Jacobiano:

$$
\boxed{
\operatorname{rank}(J)<\operatorname{rank}_{\max}
}
$$

Por tanto:

$$
\boxed{
\text{Alto riesgo dinámico}
\neq
\text{Alto riesgo de singularidad}
}
$$

Ambos fenómenos deben analizarse por separado: la dinámica determina el esfuerzo de los actuadores, mientras que el Jacobiano determina la movilidad cartesiana instantánea.
