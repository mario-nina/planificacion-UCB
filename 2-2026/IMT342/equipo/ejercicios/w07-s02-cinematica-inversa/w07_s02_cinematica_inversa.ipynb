### Enunciado

Clasifique los siguientes escenarios indicando si requieren **cinemática directa (FK)**, **cinemática inversa (IK)** o **ninguna de las dos**:

1. Dados $q_1,\ldots,q_6$, determinar la pose de la brida.
2. Dada una pose deseada para la brida, encontrar todas las combinaciones articulares admisibles.
3. Dada ${}^{0}T_{6}$, computar matemáticamente ${}^{6}T_{0}$.
4. Dada una posición objetivo para un brazo 2R, determinar las configuraciones de codo arriba y codo abajo.

### Desarrollo

#### 1. Dados $q_1,\ldots,q_6$, determinar la pose de la brida

Corresponde a **cinemática directa (FK)**.

Se conocen las variables articulares:

$$
\mathbf{q} =
\begin{bmatrix}
q_1 & q_2 & q_3 & q_4 & q_5 & q_6
\end{bmatrix}^{T}
$$

y se busca obtener la pose de la brida:

$$
{}^{0}T_{6} = FK(\mathbf{q})
$$

Por tanto:

$$
\boxed{\text{Variables articulares} \rightarrow \text{Pose} = FK}
$$


#### 2. Dada una pose deseada para la brida, encontrar todas las combinaciones articulares admisibles

Corresponde a **cinemática inversa (IK)**.

En este caso se conoce la pose deseada:

$$
{}^{0}T_{6}^{d}
$$

y se buscan las configuraciones articulares que producen dicha pose:

$$
\mathbf{q}^{(k)} = IK\left({}^{0}T_{6}^{d}\right)
$$

donde $k$ representa las distintas soluciones posibles.

Por tanto:

$$
\boxed{\text{Pose deseada} \rightarrow \text{Variables articulares} = IK}
$$


#### 3. Dada ${}^{0}T_{6}$, computar matemáticamente ${}^{6}T_{0}$

Corresponde a **ninguna de las dos**.

No se está resolviendo la configuración articular ni calculando la pose a partir de las articulaciones. Simplemente se está invirtiendo una transformación homogénea:

$$
{}^{6}T_{0}
=
\left({}^{0}T_{6}\right)^{-1}
$$

Si

$$
{}^{0}T_{6}
=
\begin{bmatrix}
R & p\\
0 & 1
\end{bmatrix},
$$

entonces

$$
{}^{6}T_{0}
=
\begin{bmatrix}
R^{T} & -R^{T}p\\
0 & 1
\end{bmatrix}
$$

Por tanto, es una **operación sobre una transformación homogénea**, no un problema de FK ni de IK.


#### 4. Dada una posición objetivo para un brazo 2R, determinar las configuraciones de codo arriba y codo abajo

Corresponde a **cinemática inversa (IK)**.

Se conoce una posición cartesiana deseada:

$$
p_d =
\begin{bmatrix}
x_d\\
y_d
\end{bmatrix}
$$

y se buscan los valores articulares:

$$
(q_1,q_2)
$$

que permiten alcanzar esa posición.

Una misma posición puede admitir más de una solución, por ejemplo:

- configuración de **codo arriba**;
- configuración de **codo abajo**.

Por tanto:

$$
\boxed{\text{Posición objetivo} \rightarrow \text{Configuraciones articulares} = IK}
$$


### Resumen

| Inciso | Clasificación |
|---|---|
| 1 | FK |
| 2 | IK |
| 3 | Ninguna |
| 4 | IK |

## Ejercicio 2 — Multiplicidad y Alcance

### Enunciado

Para un brazo planar de dos enlaces con

$$
a_1=1,\qquad a_2=1,
$$

el rango de alcance admisible es

$$
|a_1-a_2|
\leq
\sqrt{x^2+y^2}
\leq
a_1+a_2.
$$

Clasificar los siguientes objetivos como **interior alcanzable (con multiplicidad)**,
**límite singular** o **inalcanzable**:

- $P=(1,1)$
- $P=(2,0)$
- $P=(3,0)$
- $P=(0,0)$

### Desarrollo

La distancia desde la base del manipulador hasta el punto objetivo es

$$
r=\sqrt{x^2+y^2}.
$$

Como $a_1=a_2=1$:

$$
|1-1|\leq r\leq1+1
$$

$$
\boxed{0\leq r\leq2}
$$

Un punto con

$$
0<r<2
$$

se encuentra en el interior del espacio alcanzable y normalmente admite las dos
configuraciones de cinemática inversa: **codo arriba** y **codo abajo**.

#### 1. $P=(1,1)$

$$
r=\sqrt{1^2+1^2}
=\sqrt{2}
\approx1.414
$$

Como

$$
0<\sqrt{2}<2,
$$

el punto se encuentra en el interior del espacio alcanzable.

Por tanto:

$$
\boxed{P=(1,1):\ \text{Interior alcanzable con multiplicidad}}
$$

Existen dos configuraciones geométricas principales para alcanzar el punto:
**codo arriba** y **codo abajo**.


#### 2. $P=(2,0)$

$$
r=\sqrt{2^2+0^2}=2
$$

El punto se encuentra exactamente en el límite exterior:

$$
r=a_1+a_2=2.
$$

En esta posición los dos eslabones quedan completamente extendidos y las dos
soluciones de codo arriba/codo abajo colapsan en una sola configuración geométrica.

Por tanto:

$$
\boxed{P=(2,0):\ \text{Límite singular}}
$$


#### 3. $P=(3,0)$

$$
r=\sqrt{3^2+0^2}=3
$$

Como

$$
3>2,
$$

el objetivo está fuera del alcance máximo del manipulador.

Por tanto:

$$
\boxed{P=(3,0):\ \text{Inalcanzable}}
$$


#### 4. $P=(0,0)$

$$
r=\sqrt{0^2+0^2}=0
$$

El punto se encuentra exactamente en el límite interior:

$$
r=|a_1-a_2|=0.
$$

Para alcanzarlo, los dos eslabones deben quedar plegados uno sobre el otro,
correspondiendo a una configuración singular.

Por tanto:

$$
\boxed{P=(0,0):\ \text{Límite singular}}
$$

**Observación:** debido a que en este caso particular $a_1=a_2$, con sólo la
posición del extremo especificada el brazo puede plegarse sobre el origen con
$q_2=\pi$ para distintas orientaciones de $q_1$. Por ello, aunque el punto
pertenece al límite singular, no debe interpretarse literalmente como una única
pareja articular posible.


### Resumen

| Punto | $r$ | Clasificación |
|---|---:|---|
| $(1,1)$ | $\sqrt{2}$ | Interior alcanzable con multiplicidad |
| $(2,0)$ | $2$ | Límite singular |
| $(3,0)$ | $3$ | Inalcanzable |
| $(0,0)$ | $0$ | Límite singular |

## Ejercicio 3 — Uso Analítico de `atan2`

### Enunciado

Determine el ángulo principal utilizando la función cuadrante-consciente

$$
\theta = \operatorname{atan2}(y,x)
$$

para los siguientes casos:

1. $\operatorname{atan2}(1,1)$
2. $\operatorname{atan2}(1,-1)$
3. $\operatorname{atan2}(-1,-1)$
4. $\operatorname{atan2}(-1,1)$

### Desarrollo

La función

$$
\theta = \operatorname{atan2}(y,x)
$$

utiliza de forma independiente los signos de $x$ y $y$, por lo que permite determinar correctamente el cuadrante del ángulo.

#### 1. $\operatorname{atan2}(1,1)$

El punto $(1,1)$ está en el cuadrante I.

$$
\theta = \frac{\pi}{4}
$$

$$
\boxed{\theta = 45^\circ}
$$


#### 2. $\operatorname{atan2}(1,-1)$

El punto $(-1,1)$ está en el cuadrante II.

$$
\theta = \frac{3\pi}{4}
$$

$$
\boxed{\theta = 135^\circ}
$$


#### 3. $\operatorname{atan2}(-1,-1)$

El punto $(-1,-1)$ está en el cuadrante III.

El valor principal entregado por `atan2` es

$$
\theta = -\frac{3\pi}{4}
$$

equivalente geométricamente a

$$
\theta = \frac{5\pi}{4}.
$$

Por tanto:

$$
\boxed{\theta = -135^\circ}
$$


#### 4. $\operatorname{atan2}(-1,1)$

El punto $(1,-1)$ está en el cuadrante IV.

$$
\theta = -\frac{\pi}{4}
$$

$$
\boxed{\theta = -45^\circ}
$$


```python
import numpy as np

casos = [
    (1, 1),
    (1, -1),
    (-1, -1),
    (-1, 1),
]

for y, x in casos:
    theta_rad = np.arctan2(y, x)
    theta_deg = np.degrees(theta_rad)

    print(
        f"atan2({y:2d}, {x:2d}) = "
        f"{theta_rad: .6f} rad = "
        f"{theta_deg: .1f}°"
    )
```

```python
theta_atan2 = np.degrees(np.arctan2(-1, -1))
theta_atan = np.degrees(np.arctan((-1) / (-1)))

print("atan2(-1, -1):", theta_atan2, "°")
print("atan((-1)/(-1)):", theta_atan, "°")
```

### Conclusión

La función `atan2(y,x)` debe utilizarse en cinemática inversa porque distingue correctamente el cuadrante del ángulo.

Usar solamente

$$
\tan^{-1}\left(\frac{y}{x}\right)
$$

puede producir una solución con un error de $180^\circ$ en la articulación.
