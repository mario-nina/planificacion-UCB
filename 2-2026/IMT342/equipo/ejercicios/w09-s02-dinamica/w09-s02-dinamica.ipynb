# Ejercicio 1 — Identificación de Términos Dinámicos

### Consigna

Empareje lógicamente cada término de la ecuación general de manipuladores con su significado físico. Escriba la letra en el paréntesis:

$$
\tau = M(q)\ddot{q} + C(q,\dot{q})\dot{q} + g(q)
$$

- ( ) Efectos de gravedad.
- ( ) Contribución inercial.
- ( ) Torque de actuadores requerido.
- ( ) Efectos de Coriolis y centrífugos.

**Opciones:**

- **A.** $C(q,\dot{q})\dot{q}$
- **B.** $\tau$
- **C.** $M(q)\ddot{q}$
- **D.** $g(q)$

## Resolución

La ecuación dinámica general del manipulador es:

$$
\tau = M(q)\ddot{q} + C(q,\dot{q})\dot{q} + g(q)
$$

Cada término representa una contribución física diferente al esfuerzo requerido por los actuadores.

### 1. Efectos de gravedad

Los efectos de gravedad están representados por:

$$
g(q)
$$

Por lo tanto:

$$
\boxed{\text{Efectos de gravedad} \rightarrow D}
$$

Este término representa el torque necesario para sostener el robot contra la gravedad.

---

### 2. Contribución inercial

La contribución inercial está dada por:

$$
M(q)\ddot{q}
$$

Por lo tanto:

$$
\boxed{\text{Contribución inercial} \rightarrow C}
$$

Este término representa el torque necesario para acelerar las masas del mecanismo.

---

### 3. Torque de actuadores requerido

El vector que representa el torque total requerido por los actuadores es:

$$
\tau
$$

Por lo tanto:

$$
\boxed{\text{Torque de actuadores requerido} \rightarrow B}
$$

Este es el esfuerzo total que deben proporcionar los actuadores.

---

### 4. Efectos de Coriolis y centrífugos

Los efectos de Coriolis y centrífugos están representados por:

$$
C(q,\dot{q})\dot{q}
$$

Por lo tanto:

$$
\boxed{\text{Efectos de Coriolis y centrífugos} \rightarrow A}
$$

## Resultado

El emparejamiento correcto es:

| Significado físico | Término | Respuesta |
|---|---|---|
| Efectos de gravedad | $g(q)$ | **D** |
| Contribución inercial | $M(q)\ddot{q}$ | **C** |
| Torque de actuadores requerido | $\tau$ | **B** |
| Efectos de Coriolis y centrífugos | $C(q,\dot{q})\dot{q}$ | **A** |

Por tanto:

$$
\boxed{D,\ C,\ B,\ A}
$$

en el orden en que aparecen en la consigna.

# Ejercicio 2 — Completar la Derivación 1R

### Consigna

Partiendo de las energías dadas para el modelo 1R:

$$
T = \frac{1}{2}J\dot{q}^{\,2}
$$

y

$$
V = mg\ell_c(1-\cos q)
$$

demuestre paso a paso la obtención de la ecuación de torque $\tau$ aplicando el formalismo de Euler-Lagrange:

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{q}}
\right)
-
\frac{\partial L}{\partial q}
=
\tau
$$

## Resolución

Para obtener la ecuación de torque, primero construimos el Lagrangiano a partir de la energía cinética y la energía potencial.

### 1. Construcción del Lagrangiano

El Lagrangiano se define como:

$$
L(q,\dot{q}) = T - V
$$

Sustituyendo las expresiones dadas:

$$
L =
\frac{1}{2}J\dot{q}^{\,2}
-
mg\ell_c(1-\cos q)
$$

Por tanto:

$$
\boxed{
L =
\frac{1}{2}J\dot{q}^{\,2}
-
mg\ell_c(1-\cos q)
}
$$

---

### 2. Derivada parcial respecto a $\dot{q}$

Calculamos:

$$
\frac{\partial L}{\partial \dot{q}}
$$

Como solamente el término cinético depende de $\dot{q}$:

$$
\frac{\partial L}{\partial \dot{q}}
=
J\dot{q}
$$

Ahora derivamos respecto al tiempo:

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{q}}
\right)
=
\frac{d}{dt}(J\dot{q})
$$

Suponiendo que $J$ es constante:

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{q}}
\right)
=
J\ddot{q}
$$

---

### 3. Derivada parcial respecto a $q$

Ahora calculamos:

$$
\frac{\partial L}{\partial q}
$$

El término cinético no depende de $q$, por lo que solamente derivamos el término potencial:

$$
\frac{\partial L}{\partial q}
=
\frac{\partial}{\partial q}
\left[
-mg\ell_c(1-\cos q)
\right]
$$

Como:

$$
\frac{d}{dq}(1-\cos q)=\sin q
$$

obtenemos:

$$
\frac{\partial L}{\partial q}
=
-mg\ell_c\sin q
$$

Por tanto:

$$
\boxed{
\frac{\partial L}{\partial q}
=
-mg\ell_c\sin q
}
$$

---

### 4. Aplicación de Euler-Lagrange

La ecuación de Euler-Lagrange es:

$$
\frac{d}{dt}
\left(
\frac{\partial L}{\partial \dot{q}}
\right)
-
\frac{\partial L}{\partial q}
=
\tau
$$

Sustituyendo los resultados obtenidos:

$$
J\ddot{q}
-
(-mg\ell_c\sin q)
=
\tau
$$

El doble signo negativo se convierte en positivo:

$$
J\ddot{q}
+
mg\ell_c\sin q
=
\tau
$$

Por lo tanto, la ecuación de torque del modelo 1R es:

$$
\boxed{
\tau =
J\ddot{q}
+
mg\ell_c\sin q
}
$$

---

### 5. Interpretación física

La ecuación puede separarse en dos contribuciones principales:

$$
\boxed{
\tau =
\underbrace{J\ddot{q}}_{\text{contribución inercial}}
+
\underbrace{mg\ell_c\sin q}_{\text{contribución gravitacional}}
}
$$

El primer término representa el torque necesario para producir la aceleración angular del eslabón.

El segundo término representa el torque necesario para compensar el efecto de la gravedad en la configuración $q$.

## Resultado

La ecuación dinámica obtenida mediante Euler-Lagrange es:

$$
\boxed{
\tau =
J\ddot{q}
+
mg\ell_c\sin q
}
$$

Por lo tanto, el torque requerido está compuesto por:

- **Componente inercial:** $J\ddot{q}$
- **Componente gravitacional:** $mg\ell_c\sin q$

Esta expresión coincide con la derivación presentada en las diapositivas de la Semana 9, donde también se verifica que, en reposo ($\dot q=0$ y $\ddot q=0$), queda:

$$
\tau = mg\ell_c\sin q
$$

# Ejercicio 3 — Gravedad Estática

### Consigna

Para un eslabón 1R con los siguientes parámetros:

- Masa: $m = 2\ \text{kg}$
- Distancia al centro de masa: $\ell_c = 0.4\ \text{m}$
- Gravedad: $g = 9.81\ \text{m/s}^2$

Asumiendo que el robot está en reposo estático:

$$
\dot{q}=0,
\qquad
\ddot{q}=0
$$

calcule el torque requerido para sostenerlo en las siguientes configuraciones angulares:

1. $q=0\ \text{rad}$
2. $q=\frac{\pi}{6}\ \text{rad}$
3. $q=\frac{\pi}{2}\ \text{rad}$

## Resolución

Partimos de la ecuación dinámica obtenida en el ejercicio anterior:

$$
\tau =
J\ddot{q}
+
mg\ell_c\sin q
$$

Como el robot se encuentra en reposo estático:

$$
\dot{q}=0
$$

y

$$
\ddot{q}=0
$$

Por lo tanto, la contribución inercial desaparece:

$$
J\ddot{q}=0
$$

La ecuación queda:

$$
\boxed{
\tau = mg\ell_c\sin q
}
$$

### 1. Sustitución de los parámetros

Tenemos:

$$
m=2\ \text{kg}
$$

$$
\ell_c=0.4\ \text{m}
$$

$$
g=9.81\ \text{m/s}^2
$$

Calculamos primero el factor constante:

$$
mg\ell_c
=
(2)(9.81)(0.4)
$$

$$
mg\ell_c=7.848\ \text{N·m}
$$

Por lo tanto:

$$
\boxed{
\tau=7.848\sin q
}
$$

---

### 2. Configuración $q=0$

Sustituimos:

$$
\tau=7.848\sin(0)
$$

Como:

$$
\sin(0)=0
$$

entonces:

$$
\boxed{
\tau=0\ \text{N·m}
}
$$

En esta configuración no se requiere torque gravitacional.

---

### 3. Configuración $q=\frac{\pi}{6}$

Sustituimos:

$$
\tau
=
7.848\sin\left(\frac{\pi}{6}\right)
$$

Sabemos que:

$$
\sin\left(\frac{\pi}{6}\right)=\frac{1}{2}
$$

Entonces:

$$
\tau
=
7.848\left(\frac{1}{2}\right)
$$

$$
\boxed{
\tau=3.924\ \text{N·m}
}
$$

---

### 4. Configuración $q=\frac{\pi}{2}$

Sustituimos:

$$
\tau
=
7.848\sin\left(\frac{\pi}{2}\right)
$$

Como:

$$
\sin\left(\frac{\pi}{2}\right)=1
$$

obtenemos:

$$
\boxed{
\tau=7.848\ \text{N·m}
}
$$

## Resultado

Los torques necesarios para sostener el eslabón en reposo son:

| Configuración | Torque requerido |
|---|---:|
| $q=0$ rad | $\boxed{0\ \text{N·m}}$ |
| $q=\frac{\pi}{6}$ rad | $\boxed{3.924\ \text{N·m}}$ |
| $q=\frac{\pi}{2}$ rad | $\boxed{7.848\ \text{N·m}}$ |

Se observa que el torque gravitacional depende de la configuración:

$$
\tau=mg\ell_c\sin q
$$

Por eso, aunque el robot esté completamente detenido, puede necesitar un torque para sostener su posición. El torque aumenta desde $0$ hasta $7.848\ \text{N·m}$ entre $q=0$ y $q=\frac{\pi}{2}$.

# Ejercicio 4 — Contribución Inercial

### Consigna

En un instante de tiempo, el brazo de un robot tiene una inercia de:

$$
J=0.4\ \text{kg·m}^2
$$

y debe acelerar a:

$$
\ddot{q}=3\ \text{rad/s}^2
$$

Suponiendo compensación gravitatoria perfecta en ese instante, calcule la contribución de torque puramente inercial:

$$
\tau_I
$$

## Resolución

La contribución inercial del torque está dada por el término:

$$
\tau_I=J\ddot{q}
$$

Tenemos los siguientes valores:

$$
J=0.4\ \text{kg·m}^2
$$

y

$$
\ddot{q}=3\ \text{rad/s}^2
$$

Sustituimos:

$$
\tau_I=(0.4)(3)
$$

Por lo tanto:

$$
\boxed{
\tau_I=1.2\ \text{N·m}
}
$$

### Interpretación física

Este torque es el esfuerzo necesario para producir la aceleración angular del brazo.

En este ejercicio se supone una compensación gravitatoria perfecta, por lo que estamos aislando únicamente la contribución inercial:

$$
\boxed{
\tau_I=J\ddot{q}
}
$$

No se incluye una contribución gravitacional en el resultado porque la consigna establece que esta está perfectamente compensada.

## Resultado

La contribución de torque puramente inercial es:

$$
\boxed{
\tau_I=1.2\ \text{N·m}
}
$$

Por tanto, el actuador necesita aportar **1.2 N·m de torque inercial** para producir la aceleración indicada.

# Ejercicio 6 — Misma Ruta, Diferente Tiempo

### Consigna

Explique técnicamente por qué mover el robot desde el punto A hasta el punto B a lo largo del mismo camino geométrico en **2 segundos** puede demandar un esfuerzo pico (torque) mucho mayor en el actuador en comparación con ejecutar la misma tarea en **6 segundos**.

## Resolución

La trayectoria geométrica entre los puntos A y B puede ser exactamente la misma en ambos casos. Sin embargo, el tiempo disponible para recorrerla es diferente.

La ecuación dinámica general del manipulador es:

$$
\tau =
M(q)\ddot{q}
+
C(q,\dot{q})\dot{q}
+
g(q)
$$

Por lo tanto, el torque depende no solamente de la posición $q$, sino también de la velocidad $\dot{q}$ y de la aceleración $\ddot{q}$.

### 1. Movimiento en menor tiempo

Si el robot debe realizar el mismo movimiento en solo $2$ segundos, necesita alcanzar velocidades y aceleraciones mayores.

En particular, una reducción del tiempo de ejecución implica mayores valores de:

$$
\dot{q}
$$

y, especialmente:

$$
\ddot{q}
$$

El término inercial es:

$$
M(q)\ddot{q}
$$

Por lo tanto, al aumentar la aceleración, aumenta la contribución inercial al torque.

Además, en un robot de múltiples grados de libertad, los términos de Coriolis y centrífugos dependen de la velocidad:

$$
C(q,\dot{q})\dot{q}
$$

Por lo que las velocidades mayores también pueden incrementar el esfuerzo dinámico.

### 2. Movimiento en mayor tiempo

Si el mismo recorrido se realiza en $6$ segundos, el movimiento puede ejecutarse de manera más lenta.

Esto permite utilizar menores velocidades y aceleraciones:

$$
|\dot{q}| \downarrow
$$

$$
|\ddot{q}| \downarrow
$$

Por consiguiente, normalmente disminuyen las contribuciones inerciales y las asociadas a la velocidad.

### 3. Comparación

El camino geométrico es el mismo, pero los perfiles temporales son diferentes:

$$
q(t),\quad \dot{q}(t),\quad \ddot{q}(t)
$$

Por esta razón, dos movimientos que tienen los mismos puntos inicial y final pueden requerir torques muy diferentes.

## Resultado

Realizar el mismo movimiento en **2 s** puede producir un torque pico mucho mayor que realizarlo en **6 s** porque el movimiento rápido requiere mayores velocidades y aceleraciones.

La razón principal se observa en:

$$
\boxed{
\tau =
M(q)\ddot{q}
+
C(q,\dot{q})\dot{q}
+
g(q)
}
$$

Al reducir el tiempo de ejecución:

$$
|\dot{q}| \uparrow
\qquad
|\ddot{q}| \uparrow
$$

y, por tanto, pueden aumentar considerablemente las contribuciones inerciales y dependientes de la velocidad.

Por eso:

$$
\boxed{
\text{Misma trayectoria geométrica}
\neq
\text{Mismo esfuerzo dinámico}
}
$$

# Ejercicio 7 — Singularidad vs. Dinámica

### Consigna

Evalúe la siguiente afirmación como Verdadera o Falsa y explique su razonamiento:

> "Si un robot exige un torque muy alto para moverse, significa matemáticamente que está entrando en una singularidad cinemática."

## Resolución

La afirmación es **FALSA**.

Una singularidad cinemática y un torque elevado son fenómenos diferentes.

### 1. ¿Qué define una singularidad?

Una singularidad está relacionada con el Jacobiano del robot y se define mediante una pérdida de rango:

$$
\operatorname{rank}(J(q)) < \operatorname{rank}_{\max}
$$

Esto significa que el robot pierde grados de libertad operacionales instantáneos y algunas direcciones de movimiento del efector final dejan de estar disponibles.

Por lo tanto, la singularidad es un fenómeno **cinemático**.

---

### 2. ¿Qué determina el torque?

El torque requerido por los actuadores está relacionado con la dinámica del robot:

$$
\tau =
M(q)\ddot{q}
+
C(q,\dot{q})\dot{q}
+
g(q)
$$

Por lo tanto, un torque elevado puede aparecer debido a:

- una aceleración elevada;
- grandes inercias;
- efectos de Coriolis o centrífugos;
- una configuración donde la contribución gravitacional sea grande.

Ninguna de estas condiciones implica automáticamente una pérdida de rango del Jacobiano.

---

### 3. Diferencia fundamental

La singularidad se determina mediante:

$$
\boxed{
\operatorname{rank}(J(q))
}
$$

mientras que el esfuerzo dinámico se determina mediante:

$$
\boxed{
\tau(q,\dot{q},\ddot{q})
}
$$

Son magnitudes que describen fenómenos diferentes.

Por ejemplo, un robot puede encontrarse en una configuración perfectamente no singular y, aun así, requerir un torque elevado debido a la gravedad o a una aceleración grande.

## Resultado

La afirmación es:

$$
\boxed{\text{FALSA}}
$$

Un torque elevado **no demuestra** que el robot esté entrando en una singularidad.

La diferencia fundamental es:

$$
\boxed{
\text{Singularidad}
\Longrightarrow
\operatorname{rank}(J)<\operatorname{rank}_{\max}
}
$$

mientras que:

$$
\boxed{
\text{Torque}
\Longrightarrow
\tau =
M(q)\ddot{q}
+
C(q,\dot{q})\dot{q}
+
g(q)
}
$$

Por lo tanto, para determinar si existe una singularidad debemos analizar el **Jacobiano y su rango**, no simplemente observar que el torque sea elevado.
