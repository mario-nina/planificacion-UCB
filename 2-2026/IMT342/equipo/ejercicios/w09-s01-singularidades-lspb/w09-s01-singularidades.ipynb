## Ejercicio 1 — Mapeo de Velocidad y Significado

Dado el Jacobiano evaluado y el vector de velocidades articulares:

$$
J=
\begin{bmatrix}
0 & -1\\
2 & 1
\end{bmatrix},
\qquad
\dot{q}=
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

**a)** Calcule la velocidad del efector final:

$$
\dot{x}=J\dot{q}
$$

**b)** Explique físicamente qué representa la primera columna de la matriz $J$.

### Desarrollo

La relación entre la velocidad articular y la velocidad instantánea del efector es:

$$
\dot{x}=J\dot{q}
$$

Sustituyendo los datos:

$$
\dot{x}
=
\begin{bmatrix}
0 & -1\\
2 & 1
\end{bmatrix}
\begin{bmatrix}
1\\
0
\end{bmatrix}
$$

Realizando el producto matriz-vector:

$$
\dot{x}
=
\begin{bmatrix}
0(1)+(-1)(0)\\
2(1)+1(0)
\end{bmatrix}
=
\begin{bmatrix}
0\\
2
\end{bmatrix}
$$

Por tanto:

$$
\boxed{
\dot{x}=
\begin{bmatrix}
0\\
2
\end{bmatrix}
}
$$

La primera columna de $J$,

$$
J_1=
\begin{bmatrix}
0\\
2
\end{bmatrix},
$$

representa la contribución instantánea de la junta 1 a la velocidad del efector cuando esa junta tiene velocidad unitaria y las demás juntas permanecen inmóviles.

En este caso, como

$$
\dot{q}=
\begin{bmatrix}
1\\
0
\end{bmatrix},
$$

sólo se activa la junta 1. Por eso el resultado $\dot{x}$ coincide exactamente con la primera columna del Jacobiano.

```python
import numpy as np

J = np.array([
    [0, -1],
    [2,  1]
], dtype=float)

q_dot = np.array([1, 0], dtype=float)

x_dot = J @ q_dot

print("J =")
print(J)

print("\nq_dot =", q_dot)
print("x_dot =", x_dot)

assert np.allclose(x_dot, [0, 2])
```

## Ejercicio 2 — Clasificación de Rango

Dadas las matrices:

$$
A=
\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix},
\qquad
B=
\begin{bmatrix}
1 & -2\\
-2 & 4
\end{bmatrix}
$$

Calcule analíticamente el rango de ambas matrices e indique si representan un estado regular o singular.

### Desarrollo

Para una matriz $2\times2$, el rango máximo posible es:

$$
\operatorname{rank}_{\max}=2
$$

#### Matriz $A$

$$
A=
\begin{bmatrix}
1 & 0\\
0 & 1
\end{bmatrix}
$$

Sus dos columnas son linealmente independientes. Además,

$$
\det(A)=(1)(1)-(0)(0)=1
$$

Como

$$
\det(A)\neq0,
$$

la matriz tiene rango completo:

$$
\boxed{\operatorname{rank}(A)=2}
$$

Por tanto, $A$ representa un estado **regular**.

---

#### Matriz $B$

$$
B=
\begin{bmatrix}
1 & -2\\
-2 & 4
\end{bmatrix}
$$

Observamos que la segunda columna es proporcional a la primera:

$$
\begin{bmatrix}
-2\\
4
\end{bmatrix}
=
-2
\begin{bmatrix}
1\\
-2
\end{bmatrix}
$$

Por lo tanto, las columnas son linealmente dependientes y sólo existe una dirección independiente:

$$
\boxed{\operatorname{rank}(B)=1}
$$

Como

$$
\operatorname{rank}(B)<2,
$$

la matriz representa un estado **singular**.

En resumen:

$$
\boxed{\operatorname{rank}(A)=2\;\Rightarrow\;\text{regular}}
$$

$$
\boxed{\operatorname{rank}(B)=1\;\Rightarrow\;\text{singular}}
$$

```python
import numpy as np

A = np.array([
    [1, 0],
    [0, 1]
], dtype=float)

B = np.array([
    [1, -2],
    [-2, 4]
], dtype=float)

print("Rango de A:", np.linalg.matrix_rank(A))
print("Rango de B:", np.linalg.matrix_rank(B))

print("\nDet(A):", np.linalg.det(A))
print("Det(B):", np.linalg.det(B))

assert np.linalg.matrix_rank(A) == 2
assert np.linalg.matrix_rank(B) == 1
```

## Ejercicio 3 — Criterio General vs. Cuadrado

Escriba la definición universal matemática de una singularidad cinemática mediante el rango y explique por qué la ecuación

$$
\det(J)=0
$$

es un caso especial y no la regla general.

### Desarrollo

Para un Jacobiano

$$
J(q)\in\mathbb{R}^{m\times n},
$$

el rango máximo posible es

$$
\operatorname{rank}_{\max}=\min(m,n).
$$

Una configuración del robot es **singular** cuando el Jacobiano pierde rango, es decir:

$$
\boxed{
\operatorname{rank}(J(q))<\min(m,n)
}
$$

Esta es la definición general, porque puede aplicarse tanto a matrices cuadradas como a matrices no cuadradas.

---

El criterio

$$
\det(J)=0
$$

sólo puede utilizarse cuando $J$ es una matriz cuadrada.

Por ejemplo, si

$$
J\in\mathbb{R}^{6\times6},
$$

entonces:

$$
\det(J)=0
\quad\Longleftrightarrow\quad
\operatorname{rank}(J)<6.
$$

En ese caso, el determinante permite detectar la pérdida de rango.

Sin embargo, para un Jacobiano no cuadrado, por ejemplo:

$$
J\in\mathbb{R}^{6\times7},
$$

el determinante no está definido. Aun así, el robot puede encontrarse en una configuración singular si:

$$
\operatorname{rank}(J)<6.
$$

Por tanto:

$$
\boxed{
\text{Singularidad} \iff
\operatorname{rank}(J)<\operatorname{rank}_{\max}
}
$$

mientras que

$$
\boxed{
\det(J)=0
}
$$

es únicamente un criterio equivalente cuando el Jacobiano es cuadrado.
