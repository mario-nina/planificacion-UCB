## Ejercicio 1

### Enunciado

La cinemática directa de un mecanismo simple de 1 GDL es

$$
x = 2\cos q
$$

Derive analíticamente $\dot{x}$ respecto al tiempo e identifique el Jacobiano escalar $J(q)$.

### Desarrollo
La posición depende de la variable articular \(q\):

$$
x(q)=2\cos q
$$

Como \(q=q(t)\), la posición también varía con el tiempo. Aplicando la regla de la cadena:

$$
\dot{x}
=
\frac{dx}{dq}\dot{q}
$$

Derivando la función de posición respecto a \(q\):

$$
\frac{dx}{dq}
=
\frac{d}{dq}(2\cos q)
=
-2\sin q
$$

Por tanto:

$$
\boxed{
\dot{x}=-2\sin(q)\dot{q}
}
$$

La relación general de cinemática diferencial es:

$$
\dot{x}=J(q)\dot{q}
$$

Comparando ambas expresiones:

$$
\boxed{
J(q)=-2\sin q
}
$$

## Ejercicio 6

### Enunciado

Un estudiante intenta calcular:

$$
J_{v_i}
=
z_{i-1}\times
\left(o_n-o_{i-1}\right)
$$

extrayendo $z_{i-1}$ del marco local $\{1\}$ y los orígenes $o_n$ y $o_{i-1}$ del marco base $\{0\}$.

¿Es matemáticamente correcto? Explique brevemente.

### Desarrollo

No es matemáticamente correcto utilizar directamente vectores expresados en marcos de referencia diferentes.

En la expresión:

$$
J_{v_i}
=
z_{i-1}\times
\left(o_n-o_{i-1}\right)
$$

los dos vectores que participan en el producto cruz deben estar expresados en el mismo sistema de coordenadas.

Si los orígenes están expresados en el marco base:

$$
{}^{0}o_n,
\qquad
{}^{0}o_{i-1},
$$

entonces el eje de la articulación también debe expresarse en ese mismo marco:

$$
{}^{0}z_{i-1}.
$$

La expresión consistente es:

$$
\boxed{
J_{v_i}
=
{}^{0}z_{i-1}
\times
\left(
{}^{0}o_n-{}^{0}o_{i-1}
\right)
}
$$

Por tanto, si originalmente se dispone de $z_{i-1}$ expresado en el marco $\{1\}$, primero debe transformarse al marco $\{0\}$ antes de efectuar el producto cruz.

## Ejercicio 7

### Enunciado

Se determinó que el determinante del Jacobiano de traslación del brazo 2R es:

$$
\det(J_p)=a_1a_2\sin q_2
$$

Identifique en qué ángulos de $q_2$, dentro del rango

$$
(-\pi,\pi],
$$

el manipulador pierde grados de libertad operacionales.

### Desarrollo

Una configuración es singular cuando el Jacobiano pierde rango. Para el Jacobiano de traslación $J_p$, esto ocurre cuando:

$$
\det(J_p)=0
$$

Por tanto:

$$
a_1a_2\sin q_2=0
$$

Suponiendo longitudes de eslabón no nulas:

$$
a_1\neq0,
\qquad
a_2\neq0,
$$

la condición de singularidad queda:

$$
\sin q_2=0
$$

Las soluciones generales son:

$$
q_2=k\pi,
\qquad
k\in\mathbb{Z}
$$

Dentro del intervalo solicitado:

$$
(-\pi,\pi],
$$

los valores que cumplen esta condición son:

$$
\boxed{
q_2=0
\qquad \text{y} \qquad
q_2=\pi
}
$$

El valor $-\pi$ también satisface $\sin q_2=0$, pero no pertenece al intervalo dado porque el extremo $-\pi$ está excluido.

En estas configuraciones las columnas del Jacobiano pierden independencia lineal, por lo que disminuye su rango y el manipulador pierde una dirección de movilidad instantánea en el espacio cartesiano.

## Ejercicio 8

### Enunciado

Clasifique los siguientes mapeos:

1. $T_d \rightarrow q$
2. $\dot{x}_d \rightarrow \dot{q}$

Explique por qué el segundo mapeo no elimina la necesidad del primero.

### Desarrollo

El primer mapeo:

$$
T_d \rightarrow q
$$

corresponde a un problema de cinemática inversa global.

A partir de una pose deseada del efector final $T_d$, se buscan una o más configuraciones articulares:

$$
q=
\begin{bmatrix}
q_1 &
q_2 &
\cdots &
q_n
\end{bmatrix}^{T}
$$

que permitan alcanzar dicha pose.

---

El segundo mapeo:

$$
\dot{x}_d \rightarrow \dot{q}
$$

corresponde a un problema de cinemática inversa diferencial.

La relación diferencial es:

$$
\dot{x}=J(q)\dot{q}
$$

Si el Jacobiano es cuadrado y no singular, puede escribirse:

$$
\dot{q}
=
J^{-1}(q)\dot{x}_d
$$

Por tanto, este segundo problema determina las velocidades articulares necesarias para producir una velocidad cartesiana deseada en una configuración determinada.

---

La cinemática inversa diferencial no reemplaza a la cinemática inversa global porque trabaja con velocidades instantáneas y depende de la configuración actual $q$.

Es decir:

$$
\boxed{
T_d \rightarrow q
\quad \text{resuelve una pose objetivo}
}
$$

mientras que:

$$
\boxed{
\dot{x}_d \rightarrow \dot{q}
\quad \text{resuelve un movimiento instantáneo}
}
$$

Por ello, conocer $\dot{q}$ no determina directamente qué configuración articular $q$ alcanza una pose final $T_d$.

Además, el Jacobiano depende de la configuración:

$$
J=J(q)
$$

y puede perder rango en configuraciones singulares, por lo que la inversión diferencial tampoco está disponible de la misma manera en todas las configuraciones.

## A. Reconstrucción del Centro de Muñeca

### Enunciado

La ecuación de desacoplamiento establece:

$$
p_W=p_d-d_6z_6^d
$$

donde:

$$
z_6^d=
R_d
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
$$

Explique por qué un cambio en la orientación objetivo $R_d$ altera las coordenadas físicas del centro de la muñeca $p_W$, suponiendo que:

$$
d_6\neq0
$$

### Desarrollo

El centro de muñeca se obtiene a partir de la posición deseada del efector final:

$$
p_W=p_d-d_6z_6^d
$$

El vector $z_6^d$ representa la dirección del eje $z$ del efector final expresada en el marco base y depende de la orientación deseada:

$$
z_6^d=
R_d
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
$$

Por tanto, si cambia la matriz de orientación $R_d$, también cambia la dirección del vector $z_6^d$.

Como:

$$
d_6\neq0,
$$

el término:

$$
d_6z_6^d
$$

representa un desplazamiento físico no nulo desde el centro de la muñeca hasta el punto deseado del efector.

Aunque la posición deseada $p_d$ permanezca fija, al cambiar la orientación del efector cambia la dirección de ese desplazamiento.

En consecuencia:

$$
\boxed{
R_d\ \text{cambia}
\Rightarrow
z_6^d\ \text{cambia}
\Rightarrow
p_W\ \text{cambia}
}
$$

Geométricamente, el efector puede mantener el mismo punto final $p_d$ con diferentes orientaciones, pero el centro de la muñeca debe desplazarse para conservar el offset $d_6$ entre la muñeca y el efector final.

## B. Reconstrucción de Ramas

### Enunciado

Para el brazo planar 2R:

$$
c_2=\cos q_2
=
\frac{x^2+y^2-a_1^2-a_2^2}
{2a_1a_2}
$$

Al despejar el seno se obtienen dos posibles valores:

$$
s_2=
\pm\sqrt{1-c_2^2}
$$

Explique qué representan físicamente ambos signos en el espacio real del manipulador.

### Desarrollo

A partir de:

$$
\sin^2 q_2+\cos^2 q_2=1
$$

y conociendo:

$$
c_2=\cos q_2,
$$

se obtiene:

$$
s_2=
\pm\sqrt{1-c_2^2}
$$

Los signos positivo y negativo no representan simplemente dos resultados algebraicos, sino dos configuraciones geométricas distintas del manipulador.

Para una misma posición cartesiana del efector final, el brazo planar 2R puede alcanzar el punto objetivo mediante dos posturas diferentes.

Estas dos ramas se conocen como:

$$
\boxed{
\text{codo arriba}
\qquad\text{y}\qquad
\text{codo abajo}
}
$$

En una rama se utiliza:

$$
s_2=
+\sqrt{1-c_2^2}
$$

y en la otra:

$$
s_2=
-\sqrt{1-c_2^2}
$$

Ambas ramas pueden producir la misma posición del efector final, pero corresponden a configuraciones articulares diferentes.

Por tanto:

$$
\boxed{
s_2=\pm\sqrt{1-c_2^2}
}
$$

representa la existencia de múltiples soluciones físicas de cinemática inversa para una misma posición objetivo.

## C. Justificación de la función atan2

### Enunciado

Explique por qué el uso de:

$$
q_2=\operatorname{atan2}(s_2,c_2)
$$

preserva la información de la rama geométrica de manera más confiable que calcular:

$$
q_2=
\tan^{-1}
\left(
\frac{s_2}{c_2}
\right)
$$

### Desarrollo

En la resolución de cinemática inversa conocemos por separado:

$$
s_2=\sin q_2
$$

y

$$
c_2=\cos q_2.
$$

La función:

$$
q_2=\operatorname{atan2}(s_2,c_2)
$$

utiliza simultáneamente el signo del seno y el signo del coseno.

Esto permite determinar correctamente el cuadrante en el que se encuentra el ángulo $q_2$.

En cambio, si se calcula:

$$
q_2=
\tan^{-1}
\left(
\frac{s_2}{c_2}
\right),
$$

primero se forma el cociente:

$$
\frac{s_2}{c_2}.
$$

Al realizar esta división se pierde información sobre los signos individuales de $s_2$ y $c_2$.

Por ejemplo:

$$
\frac{1}{1}=1
$$

y también:

$$
\frac{-1}{-1}=1.
$$

Sin embargo, estos dos casos corresponden a cuadrantes diferentes.

Por esta razón, la función $\operatorname{atan2}(s_2,c_2)$ permite conservar la información geométrica necesaria para distinguir correctamente las distintas ramas de cinemática inversa.

Por tanto:

$$
\boxed{
q_2=\operatorname{atan2}(s_2,c_2)
}
$$

es más confiable que:

$$
\boxed{
q_2=
\tan^{-1}
\left(
\frac{s_2}{c_2}
\right)
}
$$

porque preserva la información de cuadrante y, con ello, la rama geométrica correspondiente.

## D. Secuencia de Desacoplamiento — Pieper Roadmap

### Enunciado

Ordene lógicamente los pasos del algoritmo de resolución de cinemática inversa de un manipulador de 6 GDL mediante desacoplamiento de muñeca.

### Desarrollo

La secuencia lógica es:

**1. Conocer la pose deseada**

$$
{}^{0}T_6^d
=
\begin{bmatrix}
R_d & p_d\\
0 & 1
\end{bmatrix}
$$

Esta matriz define la posición y orientación que debe alcanzar el efector final.

---

**2. Calcular el centro de muñeca**

A partir de la pose deseada se calcula:

$$
p_W
=
p_d-d_6z_6^d
$$

con:

$$
z_6^d
=
R_d
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
$$

Esto permite separar el problema de posición del problema de orientación.

---

**3. Resolver los tres ángulos del brazo**

Utilizando la posición del centro de muñeca se resuelve:

$$
(q_1,q_2,q_3)
$$

Las primeras tres articulaciones posicionan el centro de muñeca.

---

**4. Calcular la matriz de orientación intermedia**

Con los valores obtenidos de $q_1$, $q_2$ y $q_3$, se calcula:

$$
{}^{0}R_3
=
{}^{0}R_3(q_1,q_2,q_3)
$$

Esta matriz representa la orientación alcanzada por los tres primeros eslabones.

---

**5. Calcular la orientación residual**

La orientación que todavía debe aportar la muñeca se obtiene mediante:

$$
{}^{3}R_6
=
\left({}^{0}R_3\right)^T R_d
$$

Esta operación desacopla la orientación deseada del movimiento ya producido por las tres primeras articulaciones.

---

**6. Resolver los tres ángulos de la muñeca**

A partir de:

$$
{}^{3}R_6
$$

se determinan:

$$
(q_4,q_5,q_6)
$$

Estas articulaciones completan la orientación del efector final.

---

**7. Filtrar las ramas geométricas**

Se descartan las soluciones que no cumplen los límites articulares físicos del manipulador.

---

**8. Verificar mediante Cinemática Directa**

Cada vector articular candidato:

$$
q=
\begin{bmatrix}
q_1 &
q_2 &
q_3 &
q_4 &
q_5 &
q_6
\end{bmatrix}^{T}
$$

debe introducirse nuevamente en el modelo de Cinemática Directa.

La transformación obtenida debe reproducir la pose deseada:

$$
{}^{0}T_6(q)
\approx
{}^{0}T_6^d
$$

Por tanto, el flujo completo puede resumirse como:

$$
\boxed{
T_d
\rightarrow
p_W
\rightarrow
(q_1,q_2,q_3)
\rightarrow
{}^{0}R_3
\rightarrow
{}^{3}R_6
\rightarrow
(q_4,q_5,q_6)
\rightarrow
\text{filtrado}
\rightarrow
\text{FK}
}
$$
