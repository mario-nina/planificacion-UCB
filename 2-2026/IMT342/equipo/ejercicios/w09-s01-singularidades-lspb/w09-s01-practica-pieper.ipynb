## Ejercicio 1 — Reconstrucción del Centro de Muñeca

Dada una pose objetivo

$$
{}^{0}T_{6,d}=
\begin{bmatrix}
R_d & p_d\\
0 & 1
\end{bmatrix}
$$

con

$$
p_d=
\begin{bmatrix}
0.5\\
0.2\\
0.6
\end{bmatrix},
$$

y sabiendo que el eje final deseado se obtiene como

$$
z_6^d
=
R_d
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix},
$$

demuestre que

$$
p_W=
\begin{bmatrix}
0.5-d_6\\
0.2\\
0.6
\end{bmatrix}.
$$

Explique por qué la orientación deseada modifica la posición del centro de muñeca.

### Desarrollo

Para un manipulador con muñeca esférica, el centro de muñeca se obtiene retirando de la posición deseada del efector el desplazamiento final $d_6$ en la dirección del eje $z_6$ deseado:

$$
p_W=p_d-d_6z_6^d
$$

Sustituyendo:

$$
p_W=
\begin{bmatrix}
0.5\\
0.2\\
0.6
\end{bmatrix}
-
d_6
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
$$

Distribuyendo $d_6$:

$$
p_W=
\begin{bmatrix}
0.5\\
0.2\\
0.6
\end{bmatrix}
-
\begin{bmatrix}
d_6\\
0\\
0
\end{bmatrix}
$$

Por tanto:

$$
\boxed{
p_W=
\begin{bmatrix}
0.5-d_6\\
0.2\\
0.6
\end{bmatrix}
}
$$

La orientación deseada modifica la posición del centro de muñeca porque el vector

$$
z_6^d=R_d
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
$$

depende de la matriz de orientación $R_d$.

Si cambia $R_d$, cambia la dirección espacial de $z_6^d$. Como el centro de muñeca se obtiene retrocediendo una distancia $d_6$ desde el punto deseado en esa dirección, también cambia el punto $p_W$.

Por tanto, aun manteniendo fija la posición deseada $p_d$, una orientación diferente puede producir un centro de muñeca diferente.

## Ejercicio 2 — Reconstrucción de Ramas

Si durante la resolución de la cinemática inversa se obtiene:

$$
c_3=0,
$$

determine los posibles valores de $s_3$ y explique qué representan físicamente las dos soluciones.

### Desarrollo

Sabemos que:

$$
c_3=\cos(q_3),
\qquad
s_3=\sin(q_3).
$$

Además, seno y coseno cumplen:

$$
c_3^2+s_3^2=1.
$$

Como:

$$
c_3=0,
$$

entonces:

$$
0^2+s_3^2=1
$$

$$
s_3^2=1
$$

Por tanto:

$$
\boxed{s_3=\pm1}
$$

Esto produce dos posibilidades:

$$
s_3=1
$$

o bien:

$$
s_3=-1.
$$

Cada signo representa una solución articular diferente para la misma condición geométrica.

Para recuperar posteriormente el ángulo $q_3$, no basta con conocer solamente $c_3$. Debe conservarse cada combinación posible:

$$
(c_3,s_3)=(0,1)
$$

y

$$
(c_3,s_3)=(0,-1).
$$

Físicamente, estas dos posibilidades representan dos ramas distintas de la cinemática inversa: la articulación 3 puede adoptar dos configuraciones diferentes compatibles con la condición impuesta.

Por tanto:

$$
\boxed{
(c_3,s_3)=(0,1)
\quad\text{o}\quad
(c_3,s_3)=(0,-1)
}
$$
