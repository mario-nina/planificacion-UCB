# A. Reconstrucción del Centro de Muñeca

## Consigna

Asuma un manipulador con muñeca esférica y una pose final deseada dada por:

$$
{}^0T_{6,d} =
\begin{bmatrix}
R_d & p_d\\
0 & 1
\end{bmatrix}
$$

La relación del centro de muñeca se simboliza como:

$$
p_W = p_d - d_6\,{}^0z_6^d
$$

### 1.

Demuestre matricialmente cómo se obtiene el vector:

$$
{}^0z_6^d
$$

extrayéndolo directamente de la matriz $R_d$.

### 2.

Explique lógicamente por qué dos posturas del brazo totalmente distintas, por ejemplo codo-arriba y codo-abajo, pueden alcanzar físicamente el mismo punto de centro de muñeca $p_W$.

## Resolución

### A.1 Extracción de ${}^0z_6^d$

La matriz de rotación deseada se puede escribir como:

$$
R_d =
\begin{bmatrix}
{}^0x_6^d & {}^0y_6^d & {}^0z_6^d
\end{bmatrix}
$$

Cada columna de una matriz de rotación representa un eje del marco final expresado en el marco de referencia base.

Por lo tanto, el eje $z$ del marco $\{6\}$ expresado en el marco $\{0\}$ corresponde directamente a la tercera columna de $R_d$.

Si:

$$
R_d =
\begin{bmatrix}
r_{11} & r_{12} & r_{13}\\
r_{21} & r_{22} & r_{23}\\
r_{31} & r_{32} & r_{33}
\end{bmatrix}
$$

entonces:

$$
\boxed{
{}^0z_6^d =
\begin{bmatrix}
r_{13}\\
r_{23}\\
r_{33}
\end{bmatrix}
}
$$

Por lo tanto, el centro de muñeca se obtiene mediante:

$$
\boxed{
p_W = p_d-d_6
\begin{bmatrix}
r_{13}\\
r_{23}\\
r_{33}
\end{bmatrix}
}
$$

### A.2 Existencia de diferentes posturas del brazo

Una misma posición del centro de muñeca $p_W$ puede alcanzarse mediante diferentes configuraciones articulares.

En particular, un brazo puede adoptar una configuración de:

- codo arriba;
- codo abajo.

Aunque las posiciones articulares sean diferentes, ambas configuraciones pueden llevar el centro de muñeca exactamente al mismo punto del espacio.

Esto ocurre porque la cinemática inversa puede tener múltiples soluciones o ramas geométricas para una misma posición cartesiana.

Por lo tanto:

$$
\boxed{
\text{mismo }p_W
\;\not\Rightarrow\;
\text{una única configuración articular}
}
$$

Las diferentes configuraciones representan distintas ramas de la solución de cinemática inversa.

# B. Reconstrucción de Ramas (Branches)

## Consigna

De la derivación de cinemática inversa planar se obtiene:

$$
c_2 =
\frac{x^2+y^2-a_1^2-a_2^2}
{2a_1a_2}
$$

Para encontrar el ángulo, se despeja:

$$
s_2 = \pm\sqrt{1-c_2^2}
$$

### 1.

Explique el significado geométrico y físico de los signos $\pm$ presentes en la ecuación de $s_2$.

¿Qué pasaría con el manipulador en el espacio físico si uno de esos signos matemáticos se ignora sistemáticamente?

## Resolución

Partimos de:

$$
c_2=\cos(q_2)
$$

y de la identidad trigonométrica:

$$
\sin^2(q_2)+\cos^2(q_2)=1
$$

Por lo tanto:

$$
s_2^2=1-c_2^2
$$

y al extraer la raíz:

$$
\boxed{
s_2=\pm\sqrt{1-c_2^2}
}
$$

Los dos signos representan dos valores posibles de $\sin(q_2)$ para el mismo valor de $\cos(q_2)$.

Geométricamente, estas dos posibilidades corresponden a dos configuraciones diferentes del brazo para alcanzar una misma posición del efector.

Las dos ramas pueden interpretarse como configuraciones de:

$$
\boxed{\text{codo arriba}}
$$

y:

$$
\boxed{\text{codo abajo}}
$$

Por lo tanto, el signo positivo y el signo negativo no son simplemente dos resultados matemáticos equivalentes: representan diferentes configuraciones físicas del manipulador.

Si se ignora sistemáticamente uno de los signos, se elimina una de las ramas posibles de la solución.

En consecuencia, el algoritmo de cinemática inversa dejaría de proporcionar todas las configuraciones físicamente posibles para alcanzar una determinada posición.

Por tanto:

$$
\boxed{
\text{ignorar un signo}
\Rightarrow
\text{perder una rama de solución}
}
$$

# C. Razonamiento atan2

## Consigna

Escriba una explicación técnica clara de por qué:

$$
\operatorname{atan2}(s,c)
$$

es una función preferible y más segura que:

$$
\tan^{-1}\left(\frac{s}{c}\right)
$$

para recuperar ángulos de articulación en robótica.

Utilice la preservación de cuadrantes en su respuesta.

## Resolución

La función:

$$
\tan^{-1}\left(\frac{s}{c}\right)
$$

utiliza únicamente el cociente entre $s$ y $c$.

El problema es que el cociente:

$$
\frac{s}{c}
$$

no contiene por sí mismo suficiente información para distinguir todos los cuadrantes.

Por ejemplo, dos puntos que se encuentren en cuadrantes opuestos pueden producir el mismo valor del cociente $s/c$.

En cambio, la función:

$$
\operatorname{atan2}(s,c)
$$

recibe por separado los valores de seno y coseno y utiliza los signos de ambos para determinar correctamente el cuadrante.

Por ello, $\operatorname{atan2}$ permite recuperar el ángulo teniendo en cuenta la posición completa del punto $(c,s)$ en el plano.

En robótica esto es importante porque los ángulos articulares pueden encontrarse en diferentes cuadrantes.

Por lo tanto:

$$
\boxed{
q=\operatorname{atan2}(s,c)
}
$$

es una forma más robusta de recuperar el ángulo que:

$$
q=\tan^{-1}\left(\frac{s}{c}\right)
$$

porque conserva la información de cuadrante proporcionada por los signos de $s$ y $c$.

Además, cuando $c=0$, el uso directo de $s/c$ presenta una división por cero, mientras que $\operatorname{atan2}(s,c)$ está diseñado precisamente para trabajar con los dos componentes.

# D. Secuencia de Desacoplamiento (Brazo/Muñeca)

## Consigna

Ordene lógicamente los siguientes procesos para resolver la IK analítica global (1 al 8):

- Identificar la Pose Deseada.
- Enumerar y filtrar ramas geométricas por viabilidad.
- Despejar matriz de orientación $R_0^3$.
- Verificar los candidatos ${}^0T_6(q^{(k)})\approx{}^0T_{6,d}$ mediante FK.
- Calcular centro de muñeca $p_W$.
- Calcular $R_3^6=(R_0^3)^T R_d$ para la orientación residual.
- Resolver los tres ángulos de muñeca $(q_4,q_5,q_6)$.
- Resolver los tres ángulos de brazo $(q_1,q_2,q_3)$.

## Resolución

El orden lógico para resolver la cinemática inversa analítica mediante el desacoplamiento brazo-muñeca es el siguiente:

### 1. Identificar la pose deseada

Primero se conoce la transformación deseada:

$$
{}^0T_{6,d}
$$

que contiene la posición y orientación deseadas del efector final.

### 2. Calcular el centro de muñeca

Se calcula:

$$
p_W=p_d-d_6\,{}^0z_6^d
$$

Esto permite separar la posición del centro de muñeca del movimiento de la muñeca esférica.

### 3. Resolver los tres ángulos del brazo

A partir de $p_W$ se determinan:

$$
q_1,\ q_2,\ q_3
$$

incluyendo las diferentes ramas geométricas posibles.

### 4. Despejar la matriz de orientación del brazo

Una vez determinados $q_1,q_2,q_3$, se obtiene:

$$
R_0^3
$$

### 5. Calcular la orientación residual

Se calcula:

$$
R_3^6=(R_0^3)^T R_d
$$

Esta matriz representa la orientación que debe proporcionar la muñeca.

### 6. Resolver los tres ángulos de muñeca

A partir de $R_3^6$ se determinan:

$$
q_4,\ q_5,\ q_6
$$

### 7. Enumerar y filtrar las ramas geométricas

Se reúnen las diferentes combinaciones de las ramas del brazo y de la muñeca y se eliminan las soluciones que no sean físicamente viables.

### 8. Verificar los candidatos mediante FK

Finalmente, cada solución candidata $q^{(k)}$ debe comprobarse mediante cinemática directa:

$$
{}^0T_6(q^{(k)})\approx{}^0T_{6,d}
$$

Por lo tanto, el orden final es:

$$
\boxed{
\begin{array}{ll}
1.&\text{Identificar la pose deseada}\\
2.&\text{Calcular el centro de muñeca}\\
3.&\text{Resolver }q_1,q_2,q_3\\
4.&\text{Obtener }R_0^3\\
5.&\text{Calcular }R_3^6\\
6.&\text{Resolver }q_4,q_5,q_6\\
7.&\text{Enumerar y filtrar ramas}\\
8.&\text{Verificar mediante FK}
\end{array}
}
$$

# E. Lógica de Verificación

## Consigna

¿Por qué comprobar únicamente la posición:

$$
[x,y,z]^T
$$

a la salida del modelo matemático mediante FK es una evidencia estructuralmente insuficiente para dar como correcta una configuración candidata $q^{(k)}$?

## Resolución

La posición:

$$
p=
\begin{bmatrix}
x\\
y\\
z
\end{bmatrix}
$$

solamente describe dónde se encuentra el origen del efector final.

Sin embargo, una pose completa también contiene una orientación:

$$
R
$$

Por lo tanto, la pose completa está representada mediante:

$$
T=
\begin{bmatrix}
R&p\\
0&1
\end{bmatrix}
$$

Una configuración articular puede producir la posición correcta del efector y, al mismo tiempo, producir una orientación diferente de la deseada.

Por ello, comprobar solamente:

$$
p(q^{(k)})\approx p_d
$$

no garantiza que:

$$
R(q^{(k)})\approx R_d
$$

sea también cierto.

La verificación correcta debe comparar tanto la posición como la orientación:

$$
\boxed{
{}^0T_6(q^{(k)})\approx{}^0T_{6,d}
}
$$

De esta manera se comprueba que la configuración candidata reproduce la pose completa solicitada.

Por lo tanto:

$$
\boxed{
\text{posición correcta}
\neq
\text{pose completa correcta}
}
$$

La verificación mediante FK debe considerar ambas partes de la transformación: posición y orientación.

# Conclusión

En esta práctica se revisó el desacoplamiento cinemático utilizado en la cinemática inversa analítica de manipuladores con muñeca esférica.

Primero se reconstruyó el centro de muñeca a partir de la pose deseada:

$$
p_W=p_d-d_6\,{}^0z_6^d
$$

Luego se analizaron las ramas geométricas asociadas a los signos:

$$
s_2=\pm\sqrt{1-c_2^2}
$$

y la recuperación robusta de ángulos mediante:

$$
\operatorname{atan2}(s,c)
$$

También se estableció la secuencia del desacoplamiento brazo-muñeca, separando la resolución de los tres ángulos del brazo de los tres ángulos de la muñeca.

Finalmente, se determinó que una solución de cinemática inversa debe verificarse mediante la pose completa:

$$
{}^0T_6(q^{(k)})\approx{}^0T_{6,d}
$$

y no solamente mediante la posición del efector final.
