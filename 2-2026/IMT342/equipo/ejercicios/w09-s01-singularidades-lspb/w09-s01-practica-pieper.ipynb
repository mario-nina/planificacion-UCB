## Ejercicio 1 — Reconstrucción del Centro de Muñeca

Dada una pose objetivo

$$
{}^{0}T_{6,d}=
\begin{bmatrix}
R_d & p_d\\
0 & 1
\end{bmatrix}
$$

con

$$
p_d=
\begin{bmatrix}
0.5\\
0.2\\
0.6
\end{bmatrix},
$$

y sabiendo que el eje final deseado se obtiene como

$$
z_6^d
=
R_d
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix},
$$

demuestre que

$$
p_W=
\begin{bmatrix}
0.5-d_6\\
0.2\\
0.6
\end{bmatrix}.
$$

Explique por qué la orientación deseada modifica la posición del centro de muñeca.

### Desarrollo

Para un manipulador con muñeca esférica, el centro de muñeca se obtiene retirando de la posición deseada del efector el desplazamiento final $d_6$ en la dirección del eje $z_6$ deseado:

$$
p_W=p_d-d_6z_6^d
$$

Sustituyendo:

$$
p_W=
\begin{bmatrix}
0.5\\
0.2\\
0.6
\end{bmatrix}
-
d_6
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
$$

Distribuyendo $d_6$:

$$
p_W=
\begin{bmatrix}
0.5\\
0.2\\
0.6
\end{bmatrix}
-
\begin{bmatrix}
d_6\\
0\\
0
\end{bmatrix}
$$

Por tanto:

$$
\boxed{
p_W=
\begin{bmatrix}
0.5-d_6\\
0.2\\
0.6
\end{bmatrix}
}
$$

La orientación deseada modifica la posición del centro de muñeca porque el vector

$$
z_6^d=R_d
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
$$

depende de la matriz de orientación $R_d$.

Si cambia $R_d$, cambia la dirección espacial de $z_6^d$. Como el centro de muñeca se obtiene retrocediendo una distancia $d_6$ desde el punto deseado en esa dirección, también cambia el punto $p_W$.

Por tanto, aun manteniendo fija la posición deseada $p_d$, una orientación diferente puede producir un centro de muñeca diferente.

## Ejercicio 2 — Reconstrucción de Ramas

Si durante la resolución de la cinemática inversa se obtiene:

$$
c_3=0,
$$

determine los posibles valores de $s_3$ y explique qué representan físicamente las dos soluciones.

### Desarrollo

Sabemos que:

$$
c_3=\cos(q_3),
\qquad
s_3=\sin(q_3).
$$

Además, seno y coseno cumplen:

$$
c_3^2+s_3^2=1.
$$

Como:

$$
c_3=0,
$$

entonces:

$$
0^2+s_3^2=1
$$

$$
s_3^2=1
$$

Por tanto:

$$
\boxed{s_3=\pm1}
$$

Esto produce dos posibilidades:

$$
s_3=1
$$

o bien:

$$
s_3=-1.
$$

Cada signo representa una solución articular diferente para la misma condición geométrica.

Para recuperar posteriormente el ángulo $q_3$, no basta con conocer solamente $c_3$. Debe conservarse cada combinación posible:

$$
(c_3,s_3)=(0,1)
$$

y

$$
(c_3,s_3)=(0,-1).
$$

Físicamente, estas dos posibilidades representan dos ramas distintas de la cinemática inversa: la articulación 3 puede adoptar dos configuraciones diferentes compatibles con la condición impuesta.

Por tanto:

$$
\boxed{
(c_3,s_3)=(0,1)
\quad\text{o}\quad
(c_3,s_3)=(0,-1)
}
$$

## Ejercicio 3 — Reconstrucción del Ángulo con `atan2`

Para las dos ramas obtenidas anteriormente,

$$
(c_3,s_3)=(0,1)
$$

y

$$
(c_3,s_3)=(0,-1),
$$

calcule:

$$
q_3=\operatorname{atan2}(s_3,c_3)
$$

y justifique el cuadrante correspondiente.

### Desarrollo

Para reconstruir el ángulo a partir de su seno y coseno se utiliza:

$$
q_3=\operatorname{atan2}(s_3,c_3)
$$

La función `atan2` utiliza simultáneamente el signo del seno y del coseno, por lo que permite determinar correctamente el cuadrante del ángulo.

---

#### Rama 1

Para:

$$
c_3=0,
\qquad
s_3=1,
$$

se tiene:

$$
q_3=\operatorname{atan2}(1,0)
$$

por lo tanto:

$$
\boxed{
q_3=\frac{\pi}{2}
}
$$

o, en grados:

$$
\boxed{
q_3=90^\circ
}
$$

El seno es positivo y el coseno es cero, por lo que el ángulo se encuentra sobre el eje positivo del seno, correspondiente a:

$$
q_3=\frac{\pi}{2}.
$$

---

#### Rama 2

Para:

$$
c_3=0,
\qquad
s_3=-1,
$$

se tiene:

$$
q_3=\operatorname{atan2}(-1,0)
$$

por lo tanto:

$$
\boxed{
q_3=-\frac{\pi}{2}
}
$$

o, en grados:

$$
\boxed{
q_3=-90^\circ
}
$$

El seno es negativo y el coseno es cero, por lo que el ángulo se encuentra sobre el eje negativo del seno.

---

Por tanto, las dos ramas de solución son:

$$
\boxed{
q_3^{(1)}=\frac{\pi}{2}
}
$$

$$
\boxed{
q_3^{(2)}=-\frac{\pi}{2}
}
$$

El uso de `atan2` permite distinguir correctamente ambas soluciones porque conserva la información del signo de $s_3$ y $c_3$.

```python
import numpy as np

ramas = [
    (0.0,  1.0),
    (0.0, -1.0)
]

for i, (c3, s3) in enumerate(ramas, start=1):
    q3 = np.arctan2(s3, c3)

    print(f"Rama {i}")
    print(f"q3 = {q3:.6f} rad")
    print(f"q3 = {np.degrees(q3):.1f}°")
    print()
```

## Ejercicio 4 — Validación mediante Cinemática Directa

Explique por qué cada solución candidata de la cinemática inversa debe validarse mediante:

$$
{}^{0}T_6\left(q^{(k)}\right)
\approx
{}^{0}T_{6,d}.
$$

Indique qué debe comprobarse en la posición y en la orientación del efector final.

### Desarrollo

La cinemática inversa busca encontrar uno o varios vectores articulares:

$$
q^{(k)}
=
\begin{bmatrix}
q_1^{(k)} &
q_2^{(k)} &
\cdots &
q_n^{(k)}
\end{bmatrix}^{T}
$$

capaces de producir una pose deseada del efector final:

$$
{}^{0}T_{6,d}.
$$

Una vez obtenida una solución candidata $q^{(k)}$, debe sustituirse nuevamente en el modelo de cinemática directa:

$$
{}^{0}T_6\left(q^{(k)}\right).
$$

La transformación homogénea calculada tiene la forma:

$$
{}^{0}T_6\left(q^{(k)}\right)
=
\begin{bmatrix}
R\left(q^{(k)}\right) &
p\left(q^{(k)}\right)\\
0 & 1
\end{bmatrix}.
$$

La pose deseada es:

$$
{}^{0}T_{6,d}
=
\begin{bmatrix}
R_d & p_d\\
0 & 1
\end{bmatrix}.
$$

Por tanto, la validación requiere comprobar que:

$$
p\left(q^{(k)}\right)
\approx
p_d
$$

y que:

$$
R\left(q^{(k)}\right)
\approx
R_d.
$$

Es decir, la solución candidata debe reproducir tanto la **posición** como la **orientación** deseadas del efector final.

Por ello se exige:

$$
\boxed{
{}^{0}T_6\left(q^{(k)}\right)
\approx
{}^{0}T_{6,d}
}
$$

El símbolo $\approx$ se utiliza porque en una implementación numérica pueden existir pequeñas diferencias causadas por redondeo y precisión de punto flotante.

Si una solución candidata no reproduce suficientemente la transformación objetivo, entonces no debe considerarse una solución válida de la cinemática inversa.
