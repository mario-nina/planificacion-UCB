## Ejercicio 1

### Enunciado

La cinemática directa de un mecanismo simple de 1 GDL es

$$
x = 2\cos q
$$

Derive analíticamente $\dot{x}$ respecto al tiempo e identifique el Jacobiano escalar $J(q)$.

### Desarrollo
La posición depende de la variable articular \(q\):

$$
x(q)=2\cos q
$$

Como \(q=q(t)\), la posición también varía con el tiempo. Aplicando la regla de la cadena:

$$
\dot{x}
=
\frac{dx}{dq}\dot{q}
$$

Derivando la función de posición respecto a \(q\):

$$
\frac{dx}{dq}
=
\frac{d}{dq}(2\cos q)
=
-2\sin q
$$

Por tanto:

$$
\boxed{
\dot{x}=-2\sin(q)\dot{q}
}
$$

La relación general de cinemática diferencial es:

$$
\dot{x}=J(q)\dot{q}
$$

Comparando ambas expresiones:

$$
\boxed{
J(q)=-2\sin q
}
$$

## Ejercicio 2

### Enunciado

Para una junta $i$ revoluta, en una configuración dada, se conocen los vectores expresados en el marco base:

$$
{}^{0}z_{i-1}
=
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix},
\qquad
{}^{0}o_{i-1}
=
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix},
\qquad
{}^{0}o_n
=
\begin{bmatrix}
1\\
2\\
0
\end{bmatrix}
$$

Determine matemáticamente la columna completa $J_i\in\mathbb{R}^{6}$ del Jacobiano espacial.

### Desarrollo

Para una junta revoluta, la columna del Jacobiano geométrico es:

$$
J_i=
\begin{bmatrix}
{}^{0}z_{i-1}\times
\left({}^{0}o_n-{}^{0}o_{i-1}\right)
\\[4pt]
{}^{0}z_{i-1}
\end{bmatrix}
$$

Primero calculamos el vector desde el origen de la junta hasta el efector final:

$$
{}^{0}o_n-{}^{0}o_{i-1}
=
\begin{bmatrix}
1\\
2\\
0
\end{bmatrix}
-
\begin{bmatrix}
1\\
0\\
0
\end{bmatrix}
=
\begin{bmatrix}
0\\
2\\
0
\end{bmatrix}
$$

La parte lineal es:

$$
J_{v_i}
=
{}^{0}z_{i-1}\times
\left({}^{0}o_n-{}^{0}o_{i-1}\right)
$$

$$
J_{v_i}
=
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
\times
\begin{bmatrix}
0\\
2\\
0
\end{bmatrix}
=
\begin{bmatrix}
-2\\
0\\
0
\end{bmatrix}
$$

La parte angular de una junta revoluta es el propio eje de la articulación:

$$
J_{\omega_i}
=
{}^{0}z_{i-1}
=
\begin{bmatrix}
0\\
0\\
1
\end{bmatrix}
$$

Por tanto:

$$
\boxed{
J_i=
\begin{bmatrix}
-2\\
0\\
0\\
0\\
0\\
1
\end{bmatrix}
}
$$

## Ejercicio 3

### Enunciado

Para una junta $i$ prismática que se desplaza a lo largo del eje local $Y$, cuyo vector dirección expresado en el marco base actual es:

$$
{}^{0}z_{i-1}
=
\begin{bmatrix}
0\\
1\\
0
\end{bmatrix}
$$

Escriba la columna $J_i\in\mathbb{R}^{6}$ correspondiente.

### Desarrollo

Para una junta prismática, la columna del Jacobiano geométrico es:

$$
J_i=
\begin{bmatrix}
{}^{0}z_{i-1}\\
0
\end{bmatrix}
$$

A diferencia de una junta revoluta, una junta prismática produce únicamente velocidad lineal en la dirección de su eje de desplazamiento y no genera velocidad angular.

Por tanto, la parte lineal es:

$$
J_{v_i}
=
{}^{0}z_{i-1}
=
\begin{bmatrix}
0\\
1\\
0
\end{bmatrix}
$$

y la parte angular es:

$$
J_{\omega_i}
=
\begin{bmatrix}
0\\
0\\
0
\end{bmatrix}
$$

Finalmente:

$$
\boxed{
J_i=
\begin{bmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{bmatrix}
}
$$
